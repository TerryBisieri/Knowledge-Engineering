
# Knowledge Engineering: Autonomous Trailer Navigation Agent

## Step 1: The Task

The task is to develop a knowledge-based autonomous trailer agent
that can make navigation decisions using information about
Kenya's highway network.

**Goal:** Enable the trailer to navigate safely from a starting
town to a destination while minimizing the total travel distance.

The agent should:
- Identify its current location and destination.
- Access stored knowledge about towns and road connections.
- Determine available routes between towns.
- Calculate the total distance of possible routes.
- Recommend the shortest available route.
- Provide navigation instructions to the user.



## Step 2: Relevant Knowledge

The knowledge required by the trailer includes the names of
major towns, road connections, distances between towns,
and navigation rules.

The road network and distances are based on the highway
network developed in Practical Exercise 3. Major towns
were selected instead of smaller towns to simplify the
knowledge representation.

All road distances are represented in kilometres (km).
The roads are considered bidirectional, meaning the trailer
can travel between connected towns in either direction.

No external APIs or web scraping tools are integrated
into this implementation. The road information is
manually represented in the knowledge base.



## Step 3: The Vocabulary

The vocabulary defines the predicates and constants used
by the autonomous trailer's knowledge base.

### Predicates
- Town(x): Represents a town in the highway network.
- Road(x,y): Represents a road connection between two towns.
- Distance(x,y,d): Represents the distance between towns.
- CurrentLocation(x): Represents the trailer's current location.
- Destination(x): Represents the target destination.
- Route(x,y): Represents a possible route between towns.

### Constants

The constants are the major towns selected from the
Kenyan highway network, including Nairobi, Ruiru, Thika,
Nyeri, Maai Mahiu, Naivasha, Nakuru, Narok, Bomet,
Kisii, Oyugis, Homa Bay and Kisumu.

Distances are represented as numerical values in kilometres.


```python

# Step 4: Encode General Knowledge about the Domain

# Knowledge base containing towns, road connections and distances

knowledge_base = {
    "Nairobi": {
        "Ruiru": 21.88,
        "Maai Mahiu": 42.56
    },

    "Ruiru": {
        "Thika": 17.08
    },

    "Thika": {
        "Nyeri": 70.83
    },

    "Maai Mahiu": {
        "Naivasha": 34.33,
        "Narok": 79.39
    },

    "Naivasha": {
        "Nakuru": 62.34
    },

    "Narok": {
        "Bomet": 68.18
    },

    "Bomet": {
        "Kisii": 64.69
    },

    "Kisii": {
        "Oyugis": 18.75,
        "Homa Bay": 38.66
    },

    "Oyugis": {
        "Kisumu": 45.23
    },

    "Homa Bay": {
        "Kisumu": 57.78
    }
}
```

```python

# Add reverse connections to make all roads bidirectional

def make_bidirectional(kb):
    bidirectional_kb = {}

    for town, connections in kb.items():
        bidirectional_kb.setdefault(town, {})

        for next_town, distance in connections.items():
            bidirectional_kb[town][next_town] = distance

            bidirectional_kb.setdefault(next_town, {})
            bidirectional_kb[next_town][town] = distance

    return bidirectional_kb


# Update the knowledge base
knowledge_base = make_bidirectional(knowledge_base)

print("Knowledge base successfully created!")

# Display the current knowledge base

def show_kb():
    print("\n--- Current Knowledge Base ---")

    for town, connections in knowledge_base.items():
        print(f"\n{town}:")

        for next_town, distance in connections.items():
            print(f"   -> {next_town}: {distance} km")

    print("\n-------------------------------")


show_kb()

```

```python

# Step 5: Add New Facts to the Knowledge Base

def add_new_fact(town1, town2, distance):
    """Add a new bidirectional road connection to the knowledge base."""

    if distance <= 0:
        print("Distance must be greater than zero.")
        return

    knowledge_base.setdefault(town1, {})
    knowledge_base.setdefault(town2, {})

    knowledge_base[town1][town2] = distance
    knowledge_base[town2][town1] = distance

    print(f"Added road: {town1} <-> {town2} ({distance} km)")


# Example of adding a new road connection
add_new_fact("Nairobi", "Kiambu", 18.5)

def add_fact():
    """Allow the user to add a new road connection."""

    town1 = input("Enter the first town: ").strip().title()
    town2 = input("Enter the second town: ").strip().title()

    try:
        distance = float(input("Enter distance in km: "))

        add_new_fact(town1, town2, distance)

    except ValueError:
        print("Invalid distance. Please enter a number.")


# Uncomment the following line to add a road manually
# add_fact()

# View the updated knowledge base
show_kb()
```

```python

# Step 6: Encode Specific Problem Instance

current_location = input("Enter the trailer's current location: ").strip().title()

destination = input("Enter the destination: ").strip().title()

print("\n--- Navigation Problem ---")
print("Current Location:", current_location)
print("Destination:", destination)
```

```python

# Step 7: Knowledge Processing and Route Selection

import heapq

def find_shortest_route(start, goal):
    """Find the shortest route using Dijkstra's algorithm."""

    if start not in knowledge_base or goal not in knowledge_base:
        return None, float('inf')

    priority_queue = [(0, start, [start])]

    visited = set()

    while priority_queue:

        distance, current_town, path = heapq.heappop(priority_queue)

        if current_town == goal:
            return path, distance

        if current_town in visited:
            continue

        visited.add(current_town)

        for next_town, road_distance in knowledge_base[current_town].items():

            if next_town not in visited:
                new_distance = distance + road_distance

                new_path = path + [next_town]

                heapq.heappush(
                    priority_queue,
                    (new_distance, next_town, new_path)
                )

    return None, float('inf')


# Find the optimal route
route, total_distance = find_shortest_route(
    current_location, destination
)

if route:
    print("\nOptimal Route:", " -> ".join(route))
    print("Total Distance:", round(total_distance, 2), "km")
else:
    print("\nNo available route found.")
```

```python

# Step 8: Querying the Knowledge Base

def navigation_advice(start, goal):

    route, distance = find_shortest_route(start, goal)

    if route is None:
        return "No available route between the selected towns."

    return (
        f"Recommended Route: {' -> '.join(route)}\n"
        f"Total Distance: {distance:.2f} km\n"
        f"Next Town: {route[1] if len(route) > 1 else 'Already at destination'}"
    )


# Test different navigation queries

print("Query 1: Nairobi to Kisumu")
print(navigation_advice("Nairobi", "Kisumu"))

print("\nQuery 2: Kisii to Kisumu")
print(navigation_advice("Kisii", "Kisumu"))

print("\nQuery 3: Nairobi to Nyeri")
print(navigation_advice("Nairobi", "Nyeri"))
```

```python

# Step 9: Debugging the Knowledge Base

test_cases = [
    ("Nairobi", "Kisumu"),
    ("Kisii", "Kisumu"),
    ("Nairobi", "Nyeri"),
    ("Nairobi", "Mombasa"),
    ("Unknown Town", "Kisumu"),
    ("Kisumu", "Nairobi"),
    ("Ruiru", "Nyeri")
]

print("\n--- Debugging the Knowledge Base ---")

for start, goal in test_cases:

    route, distance = find_shortest_route(start, goal)

    if route:
        print(
            f"{start} -> {goal}: "
            f"{' -> '.join(route)} "
            f"({distance:.2f} km)"
        )

    else:
        print(f"{start} -> {goal}: No available route found.")
```
